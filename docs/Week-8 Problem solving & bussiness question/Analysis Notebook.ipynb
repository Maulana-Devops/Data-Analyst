# Problem Analysis

Notebook dokumentasi identifikasi masalah,
data evidence, potential causes, insights,
dan problem tree.


## Business Problem

Perusahaan mengalami penurunan penjualan pada periode tertentu
sehingga diperlukan analisis untuk mengetahui pola penurunan
dan faktor-faktor yang berpotensi berkontribusi terhadap
perubahan performa penjualan.


## Main Problem

Performa penjualan berfluktuasi dan belum stabil.

Sales terendah:
February 2023
= 6,584.00

Penurunan terbesar:
February 2023
= -8.82%


## Identified Problems

1. Penjualan mengalami fluktuasi antarperiode.
2. Kontribusi penjualan terkonsentrasi pada segmen tertentu.
3. Kualitas data masih memiliki beberapa ketidaksesuaian.


## Potential Causes

1. Perubahan performa transaksi antarperiode.
2. Perbedaan kontribusi produk.
3. Ketergantungan pada lokasi transaksi tertentu.
4. Perbedaan kontribusi metode pembayaran.
5. Ketidakkonsistenan kualitas data transaksi.


## Data Evidence

1. Sales terendah terjadi pada February 2023.
2. Penurunan terbesar terjadi pada February.
3. SALAD memiliki kontribusi sales terbesar.
4. TAKEAWAY memiliki kontribusi sales terbesar.
5. DIGITAL WALLET memiliki kontribusi sales terbesar.
6. 460 transaksi memiliki Transaction_Date missing.
7. 636 transaksi memiliki Item UNKNOWN/ERROR.
8. 696 transaksi memiliki Location UNKNOWN/ERROR.
9. 1194 transaksi memiliki sales mismatch.


## Key Insights

1. Masalah utama lebih tepat digambarkan sebagai fluktuasi
   sales antarbulan daripada penurunan terus-menerus.
2. Distribusi sales tidak merata pada produk, lokasi,
   dan metode pembayaran.
3. Kualitas data perlu diperhatikan sebelum pengambilan
   keputusan bisnis lebih lanjut.


## Problem Tree

CORE PROBLEM:
Performa penjualan berfluktuasi dan belum stabil.

DIRECT CAUSES:
- Perubahan performa penjualan antarperiode.
- Kontribusi penjualan tidak merata antarproduk.
- Ketergantungan pada lokasi transaksi tertentu.
- Perbedaan kontribusi metode pembayaran.

UNDERLYING CAUSES:
- Perubahan jumlah dan aktivitas transaksi antarperiode.
- Distribusi kontribusi produk berbeda.
- Aktivitas transaksi terkonsentrasi pada lokasi tertentu.
- Penggunaan metode pembayaran tidak merata.

DATA RELIABILITY FACTORS:
- Missing Transaction_Date.
- Item UNKNOWN/ERROR.
- Payment_Method UNKNOWN/ERROR.
- Location UNKNOWN/ERROR.
- Sales mismatch.

POTENTIAL IMPACT:
- Pertumbuhan sales sulit dipertahankan secara konsisten.
- Risiko ketergantungan pada segmen tertentu.
- Analisis bisnis dapat kurang akurat.


## Conclusion

Berdasarkan analisis, masalah utama adalah fluktuasi
performa penjualan. Beberapa faktor potensial telah
diidentifikasi, tetapi belum ditetapkan sebagai root cause
definitif.
