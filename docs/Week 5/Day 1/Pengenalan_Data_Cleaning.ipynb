```python
# ============================================================
# HARI SENIN: INITIAL DATA AUDIT (EDUSOFT CENTER)
# ============================================================

import pandas as pd
from google.colab import files

# 1. UPLOAD DATASET MENTAH
print("Silakan unggah dataset mentah yang dipilih dari Topik sebelumnya:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# ------------------------------------------------------------
# PRAKTIK METODE PANDAS MANDATORI
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("1. EKSEKUSI METODE PANDAS MANDATORI")
print("=" * 60)

# A. info()i
print("\n--- A. METODE info() ---")
df.info()

# B. isnull()
print("\n--- B. METODE isnull().sum() ---")
print(df.isnull().sum())

# C. duplicated()
print("\n--- C. METODE duplicated().sum() ---")
print(f"Total Baris Duplikat: {df.duplicated().sum()}")

# D. describe()
print("\n--- D. METODE describe() ---")
print(df.describe(include='all'))


# ------------------------------------------------------------
# PENYUSUNAN ISI LAPORAN AUDIT DATASET
# ------------------------------------------------------------
# 1. Jumlah data & Jumlah kolom
jumlah_data = len(df)
jumlah_kolom = len(df.columns)
jumlah_duplikat = df.duplicated().sum()

# 2. Tabel Missing Value, Tipe Data, & Unique Values per Kolom
audit_per_kolom = pd.DataFrame({
    'Nama Kolom': df.columns,
    'Tipe Data': df.dtypes.astype(str),
    'Missing Value': df.isnull().sum().values,
    'Persentase Missing (%)': (df.isnull().sum().values / jumlah_data * 100).round(2),
    'Jumlah Nilai Unik': [df[col].nunique() for col in df.columns]
})

# 3. Deteksi Permasalahan Otomatis
permasalahan_list = []
if jumlah_duplikat > 0:
    permasalahan_list.append(f"Ditemukan {jumlah_duplikat} baris data duplikat.")

col_missing = audit_per_kolom[audit_per_kolom['Missing Value'] > 0]['Nama Kolom'].tolist()
if col_missing:
    permasalahan_list.append(f"Ditemukan missing value pada kolom: {', '.join(col_missing)}.")

# Cek tipe data object yang mungkin perlu dikonversi ke datetime
possible_dates = [c for c in df.columns if 'date' in c.lower() or 'tanggal' in c.lower()]
if possible_dates:
    permasalahan_list.append(f"Kolom tanggal {possible_dates} masih bertipe 'object' (perlu konversi ke datetime).")

if not permasalahan_list:
    permasalahan_list.append("Tidak ditemukan masalah mayor pada audit awal.")

df_permasalahan = pd.DataFrame({'Permasalahan yang Ditemukan': permasalahan_list})


# ------------------------------------------------------------
# TAMPILAN LAPORAN PADA CONSOLE
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("2. RINGKASAN LAPORAN AUDIT DATASET")
print("=" * 60)
print(f"📄 Nama File Dataset : {filename}")
print(f"📊 Jumlah Data (Baris): {jumlah_data}")
print(f"📋 Jumlah Kolom       : {jumlah_kolom}")
print(f"👥 Jumlah Duplikat    : {jumlah_duplikat}")
print("-" * 60)
print("\n--- DETAIL PER KOLOM ---")
print(audit_per_kolom.to_string(index=False))
print("-" * 60)
print("\n--- PERMASALAHAN YANG DITEMUKAN ---")
for idx, problem in enumerate(permasalahan_list, 1):
    print(f"{idx}. {problem}")
print("=" * 60)


# ------------------------------------------------------------
# EXPORT KE FILE EXCEL (Initial Data Audit.xlsx)
# ------------------------------------------------------------
ringkasan_umum = pd.DataFrame({
    'Metrik Audit': ['Nama File', 'Jumlah Data (Baris)', 'Jumlah Kolom', 'Total Duplikat'],
    'Nilai': [filename, jumlah_data, jumlah_kolom, jumlah_duplikat]
})

excel_output = "Initial Data Audit.xlsx"

with pd.ExcelWriter(excel_output, engine='openpyxl') as writer:
    ringkasan_umum.to_excel(writer, sheet_name='Ringkasan Umum', index=False)
    audit_per_kolom.to_excel(writer, sheet_name='Audit Per Kolom', index=False)
    df_permasalahan.to_excel(writer, sheet_name='Permasalahan Ditemukan', index=False)

print(f"\n✅ File '{excel_output}' berhasil dibuat!")

# Unduh otomatis file Excel ke laptop
files.download(excel_output)
```
