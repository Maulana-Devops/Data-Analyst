```python
# ============================================================
# HARI SELASA: CLEANING MISSING VALUE & DUPLICATES
# ============================================================

import pandas as pd
from google.colab import files

# 1. Upload File Dataset Mentah (atau gunakan df dari Day 1)
print("Silakan unggah dataset mentah kamu:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

df_clean = df.copy()

# List untuk menampung tabel dokumentasi (Permasalahan & Solusi)
cleaning_log = []

# ------------------------------------------------------------
# 1. REMOVE DUPLICATE
# ------------------------------------------------------------
jumlah_duplikat = df_clean.duplicated().sum()
if jumlah_duplikat > 0:
    df_clean = df_clean.drop_duplicates()
    cleaning_log.append({
        'Permasalahan': 'Duplicate Data',
        'Solusi': 'Remove Duplicate'
    })

# ------------------------------------------------------------
# 2. HANDLING MISSING VALUES
# ------------------------------------------------------------

# A. Kolom Numerik -> Imputasi dengan Median
num_cols = ['Quantity', 'Price Per Unit', 'Total Spent']

for col in num_cols:
    if col in df_clean.columns:
        # Bersihkan karakter non-angka jika terbaca teks/string
        df_clean[col] = pd.to_numeric(
            df_clean[col].astype(str).str.extract(r'(\d+\.?\d*)')[0],
            errors='coerce'
        )
        # Cek dan isi missing value
        if df_clean[col].isnull().sum() > 0:
            df_clean[col] = df_clean[col].fillna(df_clean[col].median())
            cleaning_log.append({
                'Permasalahan': f'Missing {col}',
                'Solusi': 'Median'
            })

# B. Kolom Kategorikal & Tanggal -> Imputasi dengan Mode (Modus)
cat_cols = ['Item', 'Payment Method', 'Location', 'Transaction Date']

for col in cat_cols:
    if col in df_clean.columns:
        if df_clean[col].isnull().sum() > 0:
            mode_val = df_clean[col].mode()[0] if not df_clean[col].mode().empty else "Unknown"
            df_clean[col] = df_clean[col].fillna(mode_val)
            cleaning_log.append({
                'Permasalahan': f'Missing {col}',
                'Solusi': 'Mode'
            })

# ------------------------------------------------------------
# 3. EXPORT OUTPUT (Clean Dataset v1 & Data Cleaning Log)
# ------------------------------------------------------------
df_log = pd.DataFrame(cleaning_log)

# Export file CSV & Excel
file_clean = "Clean Dataset v1.csv"
file_log = "Data Cleaning Log.xlsx"

df_clean.to_csv(file_clean, index=False)
df_log.to_excel(file_log, index=False)

# Tampilkan hasil di console
print("\n" + "=" * 50)
print("PROSES CLEANING DAY 2 SELESAI")
print("=" * 50)
print(f"📄 Dataset Bersih : {file_clean}")
print(f"📊 Total Sisa Missing Value: {df_clean.isnull().sum().sum()}")
print("-" * 50)
print("\n--- TABEL DATA CLEANING LOG ---")
print(df_log.to_string(index=False))
print("=" * 50)

# Unduh otomatis kedua file ke laptop
files.download(file_clean)
files.download(file_log)
```
