```python
import pandas as pd
import re

def read_mysql_table(filename):
    rows = []

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            # Lewati garis pembatas tabel
            if not line or line.startswith("+"):
                continue

            # Hanya proses baris yang menggunakan |
            if "|" not in line:
                continue

            # Pisahkan berdasarkan |
            parts = [x.strip() for x in line.strip("|").split("|")]

            rows.append(parts)

    # Baris pertama dianggap sebagai header
    header = rows[0]

    # Sisanya adalah data
    data = rows[1:]

    return pd.DataFrame(data, columns=header)
```

```python
# ============================================
# PROJECT 08 - PAYMENT DATA CLEANING
# ============================================

import pandas as pd

# --------------------------------------------
# 1. Baca file tabel ASCII/MySQL
# --------------------------------------------

def read_mysql_table(filename):
    rows = []

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            # Lewati baris kosong dan garis pembatas
            if not line or line.startswith("+"):
                continue

            # Hanya ambil baris tabel
            if "|" not in line:
                continue

            parts = [
                x.strip()
                for x in line.strip("|").split("|")
            ]

            rows.append(parts)

    if not rows:
        raise ValueError(f"Tidak ada data tabel yang ditemukan pada {filename}")

    # Baris pertama = header
    header = rows[0]

    # Baris berikutnya = data
    data = rows[1:]

    return pd.DataFrame(data, columns=header)


# --------------------------------------------
# 2. Load Payment
# --------------------------------------------

df_payment = read_mysql_table("payment.txt")

# --------------------------------------------
# 3. Bersihkan nama kolom
# --------------------------------------------

df_payment.columns = (
    df_payment.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Kolom Payment:")
print(df_payment.columns.tolist())

# --------------------------------------------
# 4. Bersihkan nilai string
# --------------------------------------------

for col in df_payment.select_dtypes(
    include=["object", "string"]
).columns:
    df_payment[col] = (
        df_payment[col]
        .astype("string")
        .str.strip()
    )


# --------------------------------------------
# 5. Tampilkan struktur sebelum konversi
# --------------------------------------------

print("\nShape:")
print(df_payment.shape)

print("\nPreview:")
display(df_payment.head())


# --------------------------------------------
# 6. Konversi ID
# --------------------------------------------

if "id" in df_payment.columns:
    df_payment["id"] = pd.to_numeric(
        df_payment["id"],
        errors="coerce"
    ).astype("Int64")

if "invoice_id" in df_payment.columns:
    df_payment["invoice_id"] = pd.to_numeric(
        df_payment["invoice_id"],
        errors="coerce"
    ).astype("Int64")


# --------------------------------------------
# 7. Konversi amount
# --------------------------------------------

if "amount" in df_payment.columns:
    df_payment["amount"] = pd.to_numeric(
        df_payment["amount"],
        errors="coerce"
    )


# --------------------------------------------
# 8. Konversi payment_date
# --------------------------------------------

if "payment_date" in df_payment.columns:
    df_payment["payment_date"] = pd.to_datetime(
        df_payment["payment_date"],
        errors="coerce"
    )


# --------------------------------------------
# 9. Tambahkan Nama dan NIS
# --------------------------------------------

for col in ["nama", "nis"]:
    if col in df_payment.columns:
        df_payment = df_payment.drop(columns=[col])

df_payment.insert(
    1,
    "nama",
    "Maulana Aldi Pradana"
)

df_payment.insert(
    2,
    "nis",
    "24.012459"
)


# --------------------------------------------
# 10. Validasi relasi Payment -> Invoice
# --------------------------------------------

if (
    "invoice_id" in df_payment.columns
    and "invoice_id" in df_invoice.columns
):
    invalid_invoice = df_payment[
        ~df_payment["invoice_id"].isin(
            df_invoice["invoice_id"]
        )
    ]
else:
    invalid_invoice = pd.DataFrame()


# --------------------------------------------
# 11. HASIL AKHIR
# --------------------------------------------

print("\n" + "=" * 50)
print("PAYMENT - HASIL CLEANING")
print("=" * 50)

print("\nShape:")
print(df_payment.shape)

print("\nColumns:")
print(df_payment.columns.tolist())

print("\nMissing Values:")
print(df_payment.isnull().sum())

print("\nDuplicate Rows:")
print(df_payment.duplicated().sum())

if "id" in df_payment.columns:
    print("\nDuplicate Payment ID:")
    print(df_payment["id"].duplicated().sum())

print("\nInvoice ID yang tidak ditemukan:")
print(len(invalid_invoice))

if len(invalid_invoice) > 0:
    display(invalid_invoice)

print("\nPreview:")
display(df_payment.head())


# --------------------------------------------
# 12. Export
# --------------------------------------------

df_payment.to_csv(
    "payment_clean.csv",
    index=False
)

print("\nFile berhasil dibuat:")
print("payment_clean.csv")
```

```python
# ============================================
# PROJECT 08 - SUBSCRIPTION DATA CLEANING
# ============================================

import pandas as pd

# 1. Load Subscription
df_subscription = read_mysql_table("subscription.txt")

# 2. Bersihkan nama kolom
df_subscription.columns = (
    df_subscription.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# 3. Bersihkan kolom teks
for col in df_subscription.select_dtypes(
    include=["object", "string"]
).columns:
    df_subscription[col] = (
        df_subscription[col]
        .astype("string")
        .str.strip()
    )

# 4. Konversi ID
for col in ["id", "customer_id", "product_id"]:
    df_subscription[col] = pd.to_numeric(
        df_subscription[col],
        errors="coerce"
    ).astype("Int64")

# 5. Konversi tanggal
df_subscription["subscription_date"] = pd.to_datetime(
    df_subscription["subscription_date"],
    errors="coerce"
)

df_subscription["end_date"] = pd.to_datetime(
    df_subscription["end_date"],
    errors="coerce"
)

# 6. Tambahkan nama dan NIS
for col in ["nama", "nis"]:
    if col in df_subscription.columns:
        df_subscription = df_subscription.drop(columns=[col])

df_subscription.insert(
    1,
    "nama",
    "Maulana Aldi Pradana"
)

df_subscription.insert(
    2,
    "nis",
    "24.012459"
)

# ============================================
# VALIDASI
# ============================================

# Validasi ID Subscription
duplicate_subscription_id = df_subscription[
    df_subscription["id"].duplicated(keep=False)
]

# Validasi Customer
invalid_customer = df_subscription[
    ~df_subscription["customer_id"].isin(
        df_customer["id"]
    )
]

# Validasi Product
invalid_product = df_subscription[
    ~df_subscription["product_id"].isin(
        df_product["id"]
    )
]

# ============================================
# HASIL CLEANING
# ============================================

print("=" * 50)
print("SUBSCRIPTION - HASIL CLEANING")
print("=" * 50)

print("\nShape:")
print(df_subscription.shape)

print("\nColumns:")
print(df_subscription.columns.tolist())

print("\nMissing Values:")
print(df_subscription.isnull().sum())

print("\nDuplicate Rows:")
print(df_subscription.duplicated().sum())

print("\nDuplicate Subscription ID:")
print(df_subscription["id"].duplicated().sum())

print("\nCustomer ID yang tidak ditemukan:")
print(len(invalid_customer))

if len(invalid_customer) > 0:
    display(invalid_customer)

print("\nProduct ID yang tidak ditemukan:")
print(len(invalid_product))

if len(invalid_product) > 0:
    display(invalid_product)

print("\nPreview:")
display(df_subscription.head())

# ============================================
# EXPORT
# ============================================

df_subscription.to_csv(
    "subscription_clean.csv",
    index=False
)

print("\nFile berhasil dibuat:")
print("subscription_clean.csv")
```

```python
# ============================================
# CUSTOMER - CLEANING & TRANSFORM
# ============================================

import pandas as pd

# 1. Load Customer.txt
df_customer = read_mysql_table("customer.txt")

# 2. Bersihkan nama kolom
df_customer.columns = (
    df_customer.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Kolom awal:")
print(df_customer.columns.tolist())

# 3. Bersihkan kolom teks
for col in df_customer.select_dtypes(
    include=["object", "string"]
).columns:
    df_customer[col] = (
        df_customer[col]
        .astype("string")
        .str.strip()
    )

# 4. Konversi ID
df_customer["id"] = pd.to_numeric(
    df_customer["id"],
    errors="coerce"
).astype("Int64")

# 5. Konversi tanggal lahir
df_customer["dob"] = pd.to_datetime(
    df_customer["dob"],
    errors="coerce"
)

# ============================================
# 6. TAMBAHKAN NAMA & NIS
# ============================================

NAMA_KAMU = "Maulana Aldi Pradana"
NIS_KAMU = "24.012459"

# Hapus jika sebelumnya sudah pernah dibuat
for col in ["nama", "nis"]:
    if col in df_customer.columns:
        df_customer = df_customer.drop(columns=[col])

# Nama menjadi kolom kedua
df_customer.insert(
    1,
    "nama",
    NAMA_KAMU
)

# NIS menjadi kolom ketiga
df_customer.insert(
    2,
    "nis",
    NIS_KAMU
)

# ============================================
# 7. VALIDASI
# ============================================

duplicate_id = df_customer["id"].duplicated().sum()

# ============================================
# 8. HASIL CLEANING
# ============================================

print("=" * 50)
print("CUSTOMER - HASIL CLEANING")
print("=" * 50)

print("\nShape:")
print(df_customer.shape)

print("\nColumns:")
print(df_customer.columns.tolist())

print("\nMissing Values:")
print(df_customer.isnull().sum())

print("\nDuplicate Rows:")
print(df_customer.duplicated().sum())

print("\nDuplicate Customer ID:")
print(duplicate_id)

print("\nPreview:")
display(df_customer.head())

# ============================================
# 9. EXPORT
# ============================================

df_customer.to_csv(
    "customer_clean.csv",
    index=False
)

print("\nFile berhasil dibuat:")
print("customer_clean.csv")
```
